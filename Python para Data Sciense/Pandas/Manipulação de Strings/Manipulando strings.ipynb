MANIPULANDO STRINGS

```python
# importando a biblioteca pandas
import pandas as pd
# lendo o arquivo CSV em um DataFrame
df = pd.read_csv('C:\\Users\\eleil\\OneDrive\\Documents\\GitHub\\Estudo-e-Projetos-Alura\\Python para Data Sciense\\Pandas\\Manipulação de Strings\\banco_clientes.csv')
df
```

```python
# convertendo os valores da coluna 'nome' para letras maiúsculas
df['nome'] = df['nome'].str.upper()
df

"""O método str.upper() do pandas é utilizado para converter todos os caracteres de uma série de strings para letras maiúsculas. No exemplo acima, estamos aplicando str.upper() à coluna 'nome' do DataFrame df. Isso é útil quando queremos padronizar o formato dos dados textuais, facilitando comparações e análises posteriores. O resultado é que todos os nomes na coluna 'nome' serão exibidos em letras maiúsculas."""
```

```python
# removendo os parênteses da coluna 'telefone'
df['telefone'] = df['telefone'].str.replace("(","", regex=False).str.replace(")","", regex=False)
df

"""O metodo str.replace() do pandas é utilizado para substituir substrings em uma série de strings. No exemplo acima, estamos aplicando str.replace("(","", regex=False) e str.replace(")","", regex=False) à coluna 'telefone' do DataFrame df para remover os parênteses. O argumento regex=False indica que estamos tratando as strings literalmente, sem usar expressões regulares. Isso é útil quando queremos limpar ou padronizar os dados textuais, como números de telefone, removendo caracteres indesejados. O resultado é que todos os parênteses serão removidos dos valores na coluna 'telefone'."""
```

```python
# extraindo os códigos postais da coluna 'endereco'
df.endereco.str.extract("(\d{5}-\d{3})")


"""O metodo str.extract() do pandas é utilizado para extrair substrings de uma série de strings com base em uma expressão regular. No exemplo acima, estamos aplicando str.extract("(\d{5}-\d{3})") à coluna 'endereco' do DataFrame df para extrair os códigos postais no formato 'XXXXX-XXX'. A expressão regular (\d{5}-\d{3}) corresponde a cinco dígitos seguidos por um hífen e mais três dígitos. Isso é útil quando queremos isolar informações específicas contidas em uma string maior, facilitando a análise e manipulação dos dados. O resultado é uma nova coluna 'cep' contendo apenas os códigos postais extraídos da coluna 'endereco'."""
```

```python
df['cep'] = df.endereco.str.extract("(\d{5}-\d{3})")
df

"""O metodo str.extract() do pandas é utilizado para extrair substrings de uma série de strings com base em uma expressão regular. No exemplo acima, estamos aplicando str.extract("(\d{5}-\d{3})") à coluna 'endereco' do DataFrame df para extrair os códigos postais no formato 'XXXXX-XXX'. A expressão regular (\d{5}-\d{3}) corresponde a cinco dígitos seguidos por um hífen e mais três dígitos. Isso é útil quando queremos isolar informações específicas contidas em uma string maior, facilitando a análise e manipulação dos dados. O resultado é uma nova coluna 'cep' contendo apenas os códigos postais extraídos da coluna 'endereco'."""
```

```python
# removendo a coluna 'endereco_numero' do DataFrame
df = df.drop('endereco_numero', axis=1)
df

"""O método drop() do pandas é utilizado para remover colunas ou linhas de um DataFrame. No exemplo acima, estamos removendo a coluna 'endereco_numero' do DataFrame df, especificando axis=1 para indicar que queremos remover uma coluna (axis=0 seria para remover uma linha). Isso é útil quando queremos eliminar dados desnecessários ou irrelevantes para nossa análise. O resultado é um DataFrame atualizado sem a coluna 'endereco_numero'."""
```

```python
# removendo os códigos postais da coluna 'endereco'
df.endereco.str.replace("(\d{5}-\d{3})","", regex=True)
```

```python
# atribuindo o resultado da remoção dos códigos postais de volta à coluna 'endereco' e removendo espaços extras
df['endereco'] = df.endereco.str.replace("(\d{5}-\d{3})","", regex=True).str.strip()
df
```

```python
# Removendo a vergula da coluna 'endereco' e dividindo a coluna 'endereco' em duas novas colunas: 'rua' e 'numero'
df.endereco.str.replace(",","").str.split('\s(?=\d+$)', expand=True)

"""O método str.replace() do pandas é utilizado para substituir substrings em uma série de strings. No exemplo acima, estamos aplicando str.replace(",","") à coluna 'endereco' do DataFrame df para remover as vírgulas. Em seguida, usamos str.split('\s(?=\d+$)', expand=True) para dividir a coluna 'endereco' em duas novas colunas: 'rua' e 'numero'. A expressão regular '\s(?=\d+$)' é usada para identificar o espaço que precede o número no final da string. O argumento expand=True indica que queremos expandir o resultado em várias colunas. Isso é útil quando queremos separar informações contidas em uma única coluna para facilitar a análise e manipulação dos dados. O resultado é um DataFrame com as colunas 'rua' e 'numero' separadas.

O metodo str.split() do pandas é utilizado para dividir strings em uma série de substrings com base em um delimitador especificado. No exemplo acima, estamos usando str.split('\s(?=\d+$)', expand=True) para dividir a coluna 'endereco' do DataFrame df em duas novas colunas: 'rua' e 'numero'. A expressão regular '\s(?=\d+$)' identifica o espaço que precede o número no final da string, permitindo que a divisão ocorra nesse ponto específico. O argumento expand=True indica que queremos expandir o resultado em várias colunas. Isso é útil quando queremos separar informações contidas em uma única coluna para facilitar a análise e manipulação dos dados. O resultado é um DataFrame com as colunas 'rua' e 'numero' separadas."""
```

```python
# atribuindo as novas colunas 'rua' e 'numero' ao DataFrame
df[['rua', 'numero']] = df.endereco.str.replace(",","").str.split('\s(?=\d+$)', regex=True, expand=True)
df
```

```python
# removendo a coluna 'endereco' do DataFrame
df = df.drop('endereco', axis=1)
df

"""O metodo drop() do pandas é utilizado para remover colunas ou linhas de um DataFrame. No exemplo acima, estamos removendo a coluna 'endereco' do DataFrame df, especificando axis=1 para indicar que queremos remover uma coluna (axis=0 seria para remover uma linha). Isso é útil quando queremos eliminar dados desnecessários ou irrelevantes para nossa análise. O resultado é um DataFrame atualizado sem a coluna 'endereco'."""
```
