2) Os alunos "Alice" e "Carlos", não fazem mais parte da turma. Sendo assim, remova-os da base de dados.

```python
# separando as colunas do dataframe
# criando um dataframe com as colunas separadas
dados_temp = dados['Nome,Idade,Notas,Aprovado'].str.split(',', expand=True)
dados_temp.columns = ['Nome', 'Idade', 'Notas', 'Aprovado']
alunos_a_remover = dados_temp.query('Nome == "Alice" | Nome == "Carlos"').index # seleciona os indices dos alunos Alice e Carlos
```

```python
# removendo as linhas referentes aos alunos "Alice" e "Carlos"
dados.drop(alunos_a_remover, axis=0, inplace=True)
```

```python
# filtrando os alunos aprovados
selecao = dados_temp['Aprovado'] == 'True'
dados[selecao]
```

```python
# armazenando os dados dos alunos aprovados em um novo dataframe
alunos_aprovados = dados[selecao]
```

```python
# salvando os dados dos alunos aprovados em um arquivo csv
alunos_aprovados.to_csv('alunos_aprovados.csv', index=False)
```
