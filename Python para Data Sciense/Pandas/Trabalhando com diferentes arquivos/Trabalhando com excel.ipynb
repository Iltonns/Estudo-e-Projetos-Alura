TRABALHANDO COM ARQUIVO CSV

Um supermercado que está planejando uma liquidação de fim de ano, eles querem oferecer uma assinatura que dá 20% de desconto em todas as compras.

E o time de marketing desse supermercado entende que a melhor maneira de diminuir custos com essa campanha é analisar os dados de compras dos clientes e entender quais teriam mais chances de aderir a essa assinatura e oferecer diretamente a essas pessoas.

```python
# Importando a biblioteca pandas
import pandas as pd
# Definindo o caminho do arquivo CSV
url = 'C:\\Users\\eleil\\OneDrive\\Documents\\GitHub\\Estudo-e-Projetos-Alura\\Python para Data Sciense\\Pandas\\Trabalhando com diferentes arquivos\\dados\\superstore_data.csv'
# Lendo o arquivo CSV e armazenando em um DataFrame
dados_mercado = pd.read_csv(url)
# Exibindo as primeiras 5 linhas do DataFrame   
dados_mercado.head(5)

"""head(): Exibe as primeiras n linhas do DataFrame. O padrão é 5."""
```

```python
# Lendo um arquivo CSV com separador ponto e vírgula
url_2 = "C:\\Users\\eleil\\OneDrive\\Documents\\GitHub\\Estudo-e-Projetos-Alura\\Python para Data Sciense\\Pandas\\Trabalhando com diferentes arquivos\\dados\\superstore_data_ponto_virgula.csv"
# Lendo o arquivo CSV com separador ponto e vírgula
dados_ponto_virgula = pd.read_csv(url_2, sep=';')
# Exibindo as primeiras 5 linhas do DataFrame
dados_ponto_virgula.head(5)

'''pd.read_csv() possui diversos parâmetros que permitem customizar a leitura do arquivo CSV, como por exemplo: 
- sep: Define o separador utilizado no arquivo (padrão é vírgula).
- nrows: Define o número de linhas a serem lidas do arquivo.
- usecols: Permite selecionar colunas específicas para serem lidas do arquivo.
Esses parâmetros ajudam a adaptar a leitura do arquivo CSV conforme as necessidades específicas da análise de dados.'''
```

```python
# Lendo apenas as primeiras 5 linhas do arquivo CSV
dados_primeias_linhas = pd.read_csv(url, nrows=5)
dados_primeias_linhas

"""nrows: Define o número de linhas a serem lidas do arquivo."""
```

```python
# Lendo apenas colunas específicas do arquivo CSV
dados_selecao = pd.read_csv(url, usecols=['Id', 'Year_Birth', 'Income'])
dados_selecao.head(5)

"""usecols: Permite selecionar colunas específicas para serem lidas do arquivo."""
```

```python
# Lendo apenas colunas específicas do arquivo CSV usando índices das colunas
dados_selecao = pd.read_csv(url, usecols=[0, 1, 4])
dados_selecao.head(5)
```

```python
# Salvando o DataFrame em um novo arquivo CSV
dados_selecao.to_csv('clientes_mercado.csv', index=False)

"""index=False: Evita que o índice do DataFrame seja salvo como uma coluna no arquivo CSV."""
```

```python
# Lendo o arquivo CSV salvo anteriormente
clientes_mercado = pd.read_csv('clientes_mercado.csv')
clientes_mercado
```

```python
"""Parametros adicionais do to_csv():   
index=False: Evita que o índice do DataFrame seja salvo como uma coluna no arquivo CSV.
sep=';': Define o separador utilizado no arquivo CSV (padrão é vírgula).
header=False: Não inclui os nomes das colunas no arquivo CSV.
columns=['Id', 'Year_Birth']: Especifica quais colunas do DataFrame devem ser salvas no arquivo CSV.    
encoding='utf-8': Define a codificação do arquivo CSV (padrão é 'utf-8').   
compression='gzip': Salva o arquivo CSV em formato comprimido (gzip).
quotechar='"': Define o caractere utilizado para envolver valores de texto no arquivo CSV (padrão é aspas duplas).
date_format='%Y-%m-%d': Define o formato de data para colunas do tipo datetime no arquivo CSV.  
float_format='%.2f': Define o formato de números de ponto flutuante no arquivo CSV (neste caso, com duas casas decimais).
chunksize=1000: Salva o DataFrame em partes (chunks) de tamanho especificado, útil para grandes conjuntos de dados.
skiprows=3: indica que as três primeiras linhas do arquivo devem ser ignoradas, pois não contêm dados relevantes.
skipfooter = 9: indica que as nove últimas linhas do arquivo devem ser ignoradas, pois não contêm dados relevantes.
engine='python': o motor usado para ler o arquivo é o Python. Isso é necessário quando se usa o parâmetro skipfooter, pois o motor padrão não suporta essa opção.
usecols = 'A:C': especifica que apenas as colunas A, B e C do arquivo devem ser lidas.  
dtype={'Id': int, 'Year_Birth': int, 'Income': float}: define os tipos de dados para as colunas especificadas, garantindo que sejam lidas corretamente.
engine='python': o motor usado para ler o arquivo é o Python. Isso é necessário quando se usa o parâmetro skipfooter, pois o motor padrão não suporta essa opção.
na_values=['NA', '?']: especifica valores adicionais que devem ser tratados como valores ausentes (NaN) no DataFrame.
iterator=True: retorna um iterador que permite ler o arquivo em pedaços (chunks) de tamanho especificado.
na_filter=False: desativa a detecção automática de valores ausentes (NaN) no arquivo.
decimal=',': define o caractere usado como separador decimal no arquivo (neste caso, vírgula).
Esses parâmetros adicionais permitem maior controle sobre como o DataFrame é salvo em um arquivo CSV, adaptando-se às necessidades específicas do usuário.  """
```
