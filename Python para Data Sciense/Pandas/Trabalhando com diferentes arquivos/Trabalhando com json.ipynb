```python
# lendo o segundo arquivo JSON aninhado e armazenando em um DataFrame
dados_pacientes2 = pd.read_json('C:\\Users\\eleil\\OneDrive\\Documents\\GitHub\\Estudo-e-Projetos-Alura\\Python para Data Sciense\\Pandas\\Trabalhando com diferentes arquivos\\dados\\pacientes_2.json')
dados_pacientes2
```

```python
# normalizando os dados aninhados para criar um DataFrame plano
df_normalizado = pd.json_normalize(dados_pacientes2['Pacientes'])
df_normalizado
```

```python
# Salvando o DataFrame normalizado para um novo arquivo JSON
df_normalizado.to_json('historico_pacientes_normalizado.json')
```
