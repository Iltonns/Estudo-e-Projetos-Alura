```python
# Instalando o SQLAlchemy
%pip install sqlalchemy
# Importando as bibliotecas necessárias
import sqlalchemy
from sqlalchemy import create_engine, MetaData, Table, inspect
import pandas as pd
```

```python
# Criando a conexão com o banco de dados SQLite
engine = create_engine('sqlite:///:memory:')   
```

```python
# Importando o arquivo CSV para um DataFrame
url = r"C:\Users\eleil\OneDrive\Documents\GitHub\Estudo-e-Projetos-Alura\Python para Data Sciense\Pandas\Trabalhando com diferentes arquivos\dados\clientes_banco.csv"
# Lendo o arquivo CSV diretamente do URL
dados =pd.read_csv(url)
dados.head(5)
```

```python
# Exportando o DataFrame para uma tabela no banco de dados
dados.to_sql('clientes', engine, index=False)
```

```python
# Inspecionando o banco de dados para verificar as tabelas existentes
inspector = inspect(engine)
# Listando os nomes das tabelas no banco de dados
print(inspector.get_table_names())
```

```python
# Consultando os dados da tabela 'clientes onde a Categoria_de_renda é 'Empregado'
query = "SELECT * FROM clientes WHERE Categoria_de_renda = 'Empregado'"
```

```python
# Executando a consulta SQL onde a Categoria_de_renda é 'Empregado' e armazenando o resultado em um DataFrame
empregados = pd.read_sql(query, engine)
empregados
```

```python
# Exportando o DataFrame 'empregados' para uma nova tabela no banco de dados
empregados.to_sql('empregados', con=engine, index=False)   
```

```python
# Consultando os dados da tabela 'empregados'
pd.read_sql_table('empregados', engine)
```

```python
# Consultando apenas colunas específicas da tabela 'empregados'
pd.read_sql_table('empregados', engine, columns=['ID_Cliente', 'Grau_escolaridade', 'Rendimento_anual'])

# read_sql_table permite ler uma tabela inteira do banco de dados diretamente para um DataFrame do Pandas, facilitando a manipulação e análise dos dados armazenados no banco.
# Já o read_sql permite executar consultas SQL personalizadas, oferecendo maior flexibilidade para filtrar, agregar ou transformar os dados conforme necessário antes de carregá-los em um DataFrame.
```

```python
# Consultando os dados da tabela 'clientes' e armazenando o resultado em um DataFrame
query = "SELECT * FROM clientes"
# Executando a consulta SQL e armazenando o resultado em um DataFrame
pd.read_sql(query, engine)
```

```python
# Importando as bibliotecas necessárias para executar comandos SQL diretamente
from sqlalchemy import text
# Deletando registros da tabela 'empregados' onde o ID_Cliente é 5008804
query = 'DELETE FROM clientes WHERE ID_Cliente = 5008804'
# Executando a consulta de deleção
with engine.connect() as connection:
    # Executando a consulta SQL de deleção com segurança contra injeção de SQL  
    connection.execute(text(query))
# Confirmando as alterações no banco de dados
    connection.commit()
```

```python
# Atualizando o grau de escolaridade do cliente com ID_Cliente 5008808 para 'Ensino superior'
pd.read_sql_table('clientes', engine)   
```

```python
# Atualizando o grau de escolaridade do cliente com ID_Cliente 5008808 para 'Ensino superior'
query = "UPDATE clientes SET Grau_escolaridade = 'Ensino superior' WHERE ID_Cliente = '5008808'"
# Executando a consulta de deleção
with engine.connect() as connection:
    # Executando a consulta SQL de deleção com segurança contra injeção de SQL  
    connection.execute(text(query))
# Confirmando as alterações no banco de dados
    connection.commit()
```
