```python
"""Instalando e importando o pandas
e lendo o arquivo .csv com o pandas"""
%pip install pandas
import pandas as pd
dados = pd.read_csv("C:\\Users\\eleil\\OneDrive - Uniasselvi\\Alura\\Python para Data Sciense\\Pandas\\aluguel.csv", sep = ";")
dados.head()
```

```python
# Média simples da coluna 'Valor'
mean = dados["Valor"].mean()
mean
```

```python
# Média de valor por tipo de imóvel agrupado
dados.groupby("Tipo").mean(numeric_only=True) 
```

```python
# Média de valor por tipo de imóvel agrupado (outra forma)
dados.groupby("Tipo")["Valor"].mean(numeric_only=True) 
```

```python
# Mediana de valor por tipo de imóvel agrupado do menor para o maior
dados.groupby("Tipo")[["Valor"]].mean().sort_values('Valor')
```

```python
df_tipo_valor = dados.groupby("Tipo")[["Valor"]].mean().sort_values('Valor')
df_tipo_valor.plot.bar()
```
