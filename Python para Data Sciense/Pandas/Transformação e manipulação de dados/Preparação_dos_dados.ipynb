```python
# Importando o Pandas
import pandas as pd
# Carregando o arquivo JSON em um DataFrame
# Normalizando os dados JSON aninhados
dados = pd.json_normalize(pd.read_json('C:\\Users\\eleil\\OneDrive\\Documents\\GitHub\\Estudo-e-Projetos-Alura\\Python para Data Sciense\\Pandas\\Transformação e manipulação de dados\\dados_hospedagem.json').to_dict('records'))
dados.head()

"""O metodo pd.json_normalize() do pandas é utilizado para transformar dados JSON aninhados em um DataFrame plano. No exemplo acima, estamos lendo um arquivo JSON chamado 'dados_hospedagem.json' e convertendo-o em um DataFrame utilizando pd.read_json(). Em seguida, o método to_dict('records') é chamado para converter o DataFrame em uma lista de dicionários, onde cada dicionário representa uma linha do DataFrame. Finalmente, pd.json_normalize() é aplicado para "normalizar" os dados JSON aninhados, criando colunas separadas para cada campo aninhado e resultando em um DataFrame mais fácil de manipular e analisar.

O metodo to_dict('records') é utilizado para converter um DataFrame em uma lista de dicionários, onde cada dicionário representa uma linha do DataFrame. No exemplo acima, estamos lendo um arquivo JSON chamado 'dados_hospedagem.json' e convertendo-o em um DataFrame utilizando pd.read_json(). Em seguida, o método to_dict('records') é chamado para transformar o DataFrame em uma lista de dicionários, facilitando a manipulação dos dados JSON aninhados. Cada dicionário na lista contém os pares chave-valor correspondentes às colunas e valores do DataFrame original."""
```

```python
# Listando as colunas do DataFrame
colunas = list(dados.columns)
colunas
```

```python
# Explodindo as colunas que contêm listas em múltiplas linhas
dados = dados.explode(colunas[3:])
dados

"""O metodo explode() do pandas é utilizado para transformar colunas que contêm listas em múltiplas linhas, onde cada elemento da lista se torna uma linha separada. No exemplo acima, estamos aplicando o método explode() ao DataFrame dados, especificando as colunas a partir do índice 3 até o final da lista de colunas (colunas[3:]). Isso significa que todas as colunas a partir da quarta coluna serão "explodidas", criando novas linhas para cada elemento presente nas listas dessas colunas. O resultado é um DataFrame onde cada elemento das listas é representado em uma linha separada, facilitando a análise e manipulação dos dados."""
```

```python
# Reconfigurando o índice do DataFrame após a explosão
dados.reset_index(inplace=True, drop=True)

"""O metodo reset_index() do pandas é utilizado para redefinir o índice de um DataFrame. No exemplo acima, estamos aplicando reset_index() ao DataFrame dados após a operação de explosão realizada anteriormente. O parâmetro drop=True é utilizado para descartar o índice antigo, enquanto inplace=True indica que a modificação deve ser feita diretamente no DataFrame original, sem a necessidade de criar uma cópia. Como resultado, o DataFrame dados terá um novo índice sequencial, começando do zero, facilitando a manipulação e análise dos dados após a explosão das colunas."""
```

Tratando dados textuais

```python
dados['info_moveis.descricao_local']
```

```python
# Convertendo os valores da coluna 'info_moveis.descricao_local' para letras minúsculas e atribuindo de volta à mesma variável
dados['info_moveis.descricao_local'] = dados['info_moveis.descricao_local'].str.lower()

"""O atributo str do pandas é utilizado para acessar métodos específicos de manipulação de strings em colunas de um DataFrame. No exemplo acima, estamos utilizando str.lower() na coluna 'info_moveis.descricao_local' do DataFrame dados para converter todos os valores dessa coluna para letras minúsculas. Isso é útil quando queremos padronizar o formato dos textos, facilitando comparações e análises posteriores. Após a conversão, os valores são atribuídos de volta à mesma coluna, atualizando os dados no DataFrame."""
```

```python
# Removendo caracteres especiais da coluna 'info_moveis.descricao_local' exceto letras, números, hífens, crases e espaços e substituindo por espaço e atribuindo de volta à mesma variável
dados['info_moveis.descricao_local'] = dados['info_moveis.descricao_local'].str.replace('[^a-zA-Z0-9\-\` ]', ' ', regex=True)
dados['info_moveis.descricao_local']

"""O metodo str.replace() do pandas é utilizado para substituir ocorrências de um padrão específico em uma coluna de strings. No exemplo acima, estamos aplicando str.replace('[^a-zA-Z0-9\-\` ]', ' ', regex=True) à coluna 'info_moveis.descricao_local' do DataFrame dados. O padrão '[^a-zA-Z0-9\-\` ]' é uma expressão regular que corresponde a qualquer caractere que não seja uma letra (maiúscula ou minúscula), número, hífen, crase ou espaço. Ao utilizar esse padrão com o parâmetro regex=True, estamos substituindo todos os caracteres especiais que não correspondem a esses critérios por um espaço (' '). Isso ajuda a limpar os dados, removendo caracteres indesejados e facilitando análises posteriores. Após a substituição, os valores são atribuídos de volta à mesma coluna, atualizando os dados no DataFrame.

O regex=True indica que o padrão fornecido é uma expressão regular, permitindo correspondências mais complexas e flexíveis na substituição dos caracteres."""
```

```python
# Removendo hífens isolados da coluna 'info_moveis.descricao_local' e substituindo por espaço e atribuindo de volta à mesma variável
dados['info_moveis.descricao_local'] = dados['info_moveis.descricao_local'].str.replace('(?<!\w)-(?!\w)', ' ', regex=True)
dados['info_moveis.descricao_local']

"""Regex (ou expressão regular) é uma sequência de caracteres que define um padrão de busca em um texto. Ele é uma ferramenta poderosa e versátil que permite encontrar, substituir e manipular padrões de texto de forma eficiente. O regex é amplamente utilizado em diferentes áreas, incluindo programação, ciência de dados e processamento de texto.

Em ciência de dados, o regex é frequentemente usado para processar dados de texto brutos. Algumas dessas aplicações incluem a limpeza de dados, onde o regex pode ser usado para encontrar e substituir certos caracteres. Também pode ser usado em projetos de classificação; um exemplo é usar o regex como auxílio para encontrar padrões no texto de e-mails, o que ajuda a classificar se eles são spam ou não spam."""
```

```python
# Transformando a coluna 'info_moveis.descricao_local' em listas de palavras
dados['info_moveis.descricao_local'] = dados['info_moveis.descricao_local'].str.split()
dados['info_moveis.descricao_local']

"""O metodo str.split() do pandas é utilizado para dividir os valores de uma coluna de strings em listas de palavras, com base em um delimitador específico (por padrão, o espaço em branco). No exemplo acima, estamos aplicando str.split() à coluna 'info_moveis.descricao_local' do DataFrame dados. Isso resulta na transformação dos valores dessa coluna em listas, onde cada elemento da lista representa uma palavra separada do texto original. Essa abordagem é útil para análises de texto, como processamento de linguagem natural, onde é necessário trabalhar com palavras individuais para realizar tarefas como contagem de frequência, análise de sentimentos ou criação de modelos de aprendizado de máquina. Após a divisão, os valores são atribuídos de volta à mesma coluna, atualizando os dados no DataFrame."""
```

Trabalhando com dados de tempo

A classe datetime é um tipo de dado que representa uma data e hora específicas em Python. Ela permite trabalhar e manipular dados de ano, mês, dia, hora, minuto, segundo e microssegundo, além de dias da semana, como segunda, terça, quarta, etc. Essa classe é definida no módulo datetime e é fundamental na manipulação de datas e horários em Python. Com ela, é possível realizar operações de cálculo de diferenças entre datas, formatação de datas e horas em diferentes formatos, além de ser muito útil em análises de dados que envolvem séries temporais.

```python
# Lendo o arquivo JSON em um DataFrame e atribuindo a uma nova variável
dt_data = pd.read_json('C:\\Users\\eleil\\OneDrive\\Documents\\GitHub\\Estudo-e-Projetos-Alura\\Python para Data Sciense\\Pandas\\Transformação e manipulação de dados\\moveis_disponiveis.json')
# Exibindo as primeiras linhas do DataFrame
dt_data.head()  
```

```python
# Transformando a coluna 'data' em datetime e atribuindo de volta à mesma variável
dt_data['data'] = pd.to_datetime(dt_data['data'])
dt_data.info()

"""O metodo pd.to_datetime() do pandas é utilizado para converter uma coluna de strings ou outros tipos de dados em objetos datetime. No exemplo acima, estamos aplicando pd.to_datetime() à coluna 'data' do DataFrame dt_data, convertendo seus valores para o tipo datetime. Isso é útil quando queremos trabalhar com dados de data e hora de forma mais eficiente, permitindo operações como cálculos de diferenças entre datas, filtragem por períodos específicos e análise temporal. Após a conversão, os valores são atribuídos de volta à mesma coluna, atualizando os dados no DataFrame.

A classe datetime é uma das mais importantes e amplamente utilizadas para manipulação de datas e horas em Python. Ela representa uma data e hora específicas, combinando informações de ano, mês, dia, hora, minuto, segundo e microssegundo em um único objeto. A classe datetime oferece uma variedade de métodos e atributos que permitem realizar operações como cálculos de diferenças entre datas, formatação de datas em diferentes estilos, extração de componentes individuais (como ano ou mês) e muito mais. Além disso, a classe datetime é fundamental para trabalhar com séries temporais e análises de dados que envolvem datas e horários.

Dentro da classe datetime podemos encontrar diversos métodos úteis, como:
- dt.strftime(): Formata objetos datetime em strings com um formato específico.
- dt.strptime(): Converte strings em objetos datetime com base em um formato específico.
- dt.date(): Extrai a parte da data de um objeto datetime.
- dt.time(): Extrai a parte do tempo de um objeto datetime.
- dt.year, dt.month, dt.day: Acessa os componentes individuais de um objeto datetime.
- dt.weekday(): Retorna o dia da semana como um número inteiro (0 para segunda-feira, 6 para domingo). 
- dt.now(): Retorna o objeto datetime atual.
- dt.timedelta(): Permite realizar operações de adição e subtração de períodos de tempo.    
- dt.to_period(): Converte objetos datetime em períodos (como meses ou anos).
- dt.today(): Retorna a data atual sem a parte do tempo.
- dt.normalize(): Normaliza objetos datetime para a meia-noite do dia correspondente.
- dt.isocalendar(): Retorna uma tupla com o ano ISO, semana ISO e dia da semana ISO.
- dt.tz_localize(): Localiza um objeto datetime em um fuso horário específico.
Essas são apenas algumas das funcionalidades oferecidas pela classe datetime, que é essencial para manipulação de datas"""
```
