Existem diversas formas de tratar dados nulos com o Pandas. Algumas das principais formas são:

Remover os dados nulos: É possível remover as linhas ou colunas que possuem valores nulos utilizando o método dropna() . Esse método remove todas as linhas ou colunas que possuem pelo menos um valor nulo.

Preencher os dados nulos: Utilizando o método fillna(), podemos preencher os valores nulos com um valor específico. Além disso, também é possível utilizar argumentos específicos do método fillna() como o method=”ffill” ou method=”bfill” para preencher os valores nulos com o valor anterior ou posterior, respectivamente.

Interpolar os dados nulos: É possível utilizar o método interpolate() para preencher os valores nulos com valores interpolados, ou seja, valores calculados a partir dos valores vizinhos.

MANIPULANDO OS DADOS

Criando colunas numéricas

```python
# Lendo o arquivo CSV com pandas
dados = pd.read_csv("C:\\Users\\eleil\\OneDrive - Uniasselvi\\Alura\\Python para Data Sciense\\Pandas\\aluguel.csv", sep = ";")
dados.head()
```

```python
# Criando uma nova coluna 'Valor_por_mes' somando o valor e o condomínio
dados['Valor_por_mes'] = dados['Valor'] + dados['Condominio']  # cria uma nova coluna 'Valor_por_mes' somando o valor e o condomínio
dados.head()
```

```python
# Criando uma nova coluna 'Valor_por_ano' multiplicando o valor por mês por 12 e somando o IPTU
dados['Valor_por_ano'] = dados['Valor_por_mes'] * 12 + dados['IPTU'] # cria uma nova coluna 'Valor_por_ano' multiplicando o valor por mês por 12 e somando o IPTU
dados.head()
```

CRIANDO COLUNAS CATEGORICAS

```python
# Criando uma nova coluna 'Descricao' concatenando informações de outras colunas
dados['Descricao'] = dados['Tipo'] + ' em ' + dados['Bairro'] + ' com ' + \
                    dados['Quartos'].astype(str) + ' quarto(s) ' + 'e ' + \
                    dados['Vagas'].astype(str) + ' Vaga(s) de garagem.' 
dados.head()
```

```python
# Criando uma nova coluna 'Possui_suite' com valor booleano indicando se o imóvel possui suíte
dados['Possui_suite'] = dados['Suites'].apply(lambda x: 'Sim' if x > 0 else 'Não')  # cria uma nova coluna 'Possui_suite' com valor booleano True ou False
dados.head()
```

```python
# Salvando o DataFrame modificado em um novo arquivo CSV
dados.to_csv('Dados_completos_dev.csv', index=False, sep=';') # salva o dataframe em um arquivo csv sem o index e separador ;
```

Existem diversas maneiras de criar colunas em um DataFrame no Pandas. Alguns exemplos são:

1 - Atribuição direta de valores a uma nova coluna:

```python
import pandas as pd

df = pd.DataFrame({'A': [1, 2, 3], 'B': [4, 5, 6]})
df['C'] = [7, 8, 9]
df
```

2 - Criação de uma coluna a partir de operações entre outras colunas:

```python
import pandas as pd

df = pd.DataFrame({'A': [1, 2, 3], 'B': [4, 5, 6]})
df['C'] = df['A'] + df['B']
df
```

3 - Utilização do método assign() para criar uma nova coluna:

```python
import pandas as pd

df = pd.DataFrame({'A': [1, 2, 3], 'B': [4, 5, 6]})
df = df.assign(C=[7, 8, 9])
df
```

4 - Utilização do método apply() para aplicar uma função a uma coluna existente e criar uma nova coluna:

```python
import pandas as pd

df = pd.DataFrame({'A': [1, 2, 3], 'B': [4, 5, 6]})
df['C'] = df['A'].apply(lambda x: x * 2)
df
```
