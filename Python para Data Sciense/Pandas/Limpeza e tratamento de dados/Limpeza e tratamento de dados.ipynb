## Trabalhando com Outliers

Identificando os Outliers

```python
# Trabalhando com Outliers
# Identificando os Outliers
"""Outliers são valores que se desviam significativamente do padrão geral dos dados em um conjunto de dados. Eles podem ser causados por erros de medição, variações naturais ou eventos raros. Identificar e lidar com outliers é importante, pois eles podem distorcer análises estatísticas, afetar modelos preditivos e levar a conclusões incorretas. Métodos comuns para identificar outliers incluem o uso de gráficos de caixa (box plots), z-scores e o método do intervalo interquartil (IQR). Após a identificação, os outliers podem ser removidos, transformados ou tratados de outras maneiras, dependendo do contexto da análise.
"""

# Visualizando as esrtatísticas descritivas do DataFrame sem nulos
df_dados_sem_nulos.describe()

"""O metodo describe() do pandas é utilizado para gerar estatísticas descritivas resumidas de um DataFrame ou Série. Ele fornece informações como contagem, média, desvio padrão, valores mínimos e máximos, bem como os quartis (25%, 50% e 75%) para colunas numéricas. Isso é útil para obter uma visão geral rápida dos dados, identificar tendências e detectar possíveis outliers ou anomalias em conjuntos de dados."""
```

```python
# Instando a biblioteca seaborn
%pip install seaborn
# importando a biblioteca seaborn
import seaborn as sns
```

```python
# Visualizando os Outliers na coluna 'cliente.tempo_servico
sns.boxplot(x=df_dados_sem_nulos['cliente.tempo_servico'])

"""A biblioteca seaborn é uma biblioteca de visualização de dados em Python que fornece uma interface de alto nível para criar gráficos estatísticos atraentes e informativos. Ela é construída sobre a biblioteca Matplotlib e integra-se bem com o pandas, facilitando a criação de visualizações a partir de DataFrames. O seaborn oferece uma variedade de gráficos, como gráficos de dispersão, gráficos de barras, histogramas, gráficos de violino, gráficos de caixa (box plots) e muito mais. Além disso, ele inclui funcionalidades para personalizar estilos, cores e temas, tornando mais fácil a criação de visualizações esteticamente agradáveis e eficazes para análise exploratória de dados."""
```

```python
# Calculando os limites inferior e superior para identificar os outliers na coluna 'cliente.tempo_servico' usando o método do IQR
Q1 = df_dados_sem_nulos['cliente.tempo_servico'].quantile(.25)
Q3 = df_dados_sem_nulos['cliente.tempo_servico'].quantile(.75)
IQR = Q3 - Q1
limite_inferior = Q1 - 1.5*IQR
limite_superior = Q3 + 1.5*IQR

"""O metodo quantile() do pandas é utilizado para calcular os quantis de uma Série ou DataFrame. Um quantil é um valor que divide um conjunto de dados em partes iguais, permitindo a análise da distribuição dos dados. Por exemplo, o quantil 0.25 (Q1) representa o valor abaixo do qual 25% dos dados estão localizados, enquanto o quantil 0.75 (Q3) representa o valor abaixo do qual 75% dos dados estão localizados. O método quantile() é útil para entender a dispersão e a tendência central dos dados, bem como para identificar outliers e outras características importantes da distribuição dos dados."""
```

```python
# Identificando os outliers na coluna 'cliente.tempo_servico'
(df_dados_sem_nulos['cliente.tempo_servico'] < limite_inferior) | (df_dados_sem_nulos['cliente.tempo_servico'] > limite_superior)
```

```python
# Identificando os outliers na coluna 'cliente.tempo_servico' e atribuindo o resultado a variável outliers_index
outliers_index = (df_dados_sem_nulos['cliente.tempo_servico'] < limite_inferior) | (df_dados_sem_nulos['cliente.tempo_servico'] > limite_superior)
outliers_index
```

```python
# Visualizando os registros com outliers na coluna 'cliente.tempo_servico'
df_dados_sem_nulos[outliers_index]['cliente.tempo_servico']
```

Substituindo valores para os Otliers

```python
# Substituindo valores para os Otliers
df_sem_outliers = df_dados_sem_nulos.copy()
```

```python
# Vizualizando os valores para substituir os outliers na coluna 'cliente.tempo_servico'
df_sem_outliers[outliers_index]['cliente.tempo_servico']
```

```python
# Substituindo valores para os Otliers
df_sem_outliers.loc[outliers_index, 'cliente.tempo_servico'] = np.ceil(
    df_sem_outliers.loc[outliers_index, 'conta.cobranca.Total'] / df_sem_outliers.loc[outliers_index, 'conta.cobranca.mensal']
)

"""O metodo loc[] do pandas é utilizado para acessar um grupo de linhas e colunas em um DataFrame com base em rótulos ou uma matriz booleana. Ele permite a seleção de dados de forma eficiente, permitindo que você especifique tanto as linhas quanto as colunas que deseja acessar. A sintaxe básica é "df.loc[linhas, colunas]", onde "linhas" pode ser um rótulo, uma lista de rótulos ou uma matriz booleana, e "colunas" pode ser um rótulo, uma lista de rótulos ou uma matriz booleana. O loc[] é especialmente útil para filtrar dados, atualizar valores e realizar operações condicionais em DataFrames.

O metodo ceil() da biblioteca numpy é utilizado para arredondar valores numéricos para o próximo número inteiro mais alto. Ele retorna o menor inteiro maior ou igual ao valor fornecido. Isso é útil em situações onde você precisa garantir que os valores sejam arredondados para cima, independentemente de sua parte decimal. Por exemplo, np.ceil(4.2) retornará 5.0, e np.ceil(-3.7) retornará -3.0. O método ceil() é frequentemente utilizado em cálculos financeiros, estatísticas e outras aplicações onde o arredondamento para cima é necessário."""
```

```python
# Vizualizando os valores para substituir os outliers na coluna 'cliente.tempo_servico'
sns.boxplot(x=df_sem_outliers['cliente.tempo_servico'])

"""O metodo sns.boxplot() da biblioteca seaborn é utilizado para criar gráficos de caixa (box plots) que visualizam a distribuição de um conjunto de dados. Um box plot exibe a mediana, os quartis e os valores extremos (outliers) de uma variável numérica, permitindo uma análise rápida da dispersão e da simetria dos dados. O box plot é útil para identificar outliers, comparar distribuições entre diferentes grupos e entender a variabilidade dos dados. A sintaxe básica é sns.boxplot(x=data), onde 'data' é a série ou coluna do DataFrame que você deseja visualizar."""
```

```python
"""Existem várias técnicas para detectar outliers em um conjunto de dados. Algumas das mais comuns são: Z-score, regra dos 3 sigmas e a análise de dispersão.

Z-score:

O z-score é uma medida estatística que indica a quantos múltiplos do desvio-padrão um dado está distante da média. Para detectar outliers utilizando essa medida, basta calcular o z-score de cada dado e verificar se ele está muito distante da média. Se o z-score for maior do que um determinado limite, podemos considerar esse dado como um outlier.

Regra dos 3 sigmas:

A regra dos 3 sigmas é uma técnica que utiliza a distribuição normal para identificar outliers. Ela diz que cerca de 99,7% dos dados estarão dentro de 3 desvios padrão da média. Portanto, se um dado estiver a mais de 3 desvios padrão da média, podemos considerá-lo como um outlier.

Análise de dispersão

A análise de dispersão é uma técnica que utiliza gráficos para identificar outliers. Um gráfico comumente utilizado é o boxplot que mostra a distribuição dos dados em quartis. Os outliers são identificados como pontos fora dos limites superior e inferior do boxplot.
"""
```
