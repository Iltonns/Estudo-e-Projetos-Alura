Transformando dados em uma DataFrame

```python
# impoirtando a biblioteca json para manipulação direta do arquivo json
import json
# lendo o arquivo json diretamente
with open('C:\\Users\\eleil\\OneDrive\\Documents\\GitHub\\Estudo-e-Projetos-Alura\\Python para Data Sciense\\Pandas\\Limpeza e tratamento de dados\\dataset-telecon.json') as f:
    json_bruto = json.load(f)
json_bruto  

"""O metodo with open() é utilizado para abrir arquivos em Python de forma segura e eficiente. Ele garante que o arquivo seja corretamente fechado após o uso, mesmo que ocorra um erro durante a manipulação do arquivo. A sintaxe básica é "with open('caminho/do/arquivo', 'modo') as variavel:", onde 'caminho/do/arquivo' é o caminho para o arquivo que você deseja abrir, 'modo' especifica como o arquivo será aberto (por exemplo, 'r' para leitura, 'w' para escrita, etc.), e 'variavel' é o nome da variável que você usará para interagir com o arquivo dentro do bloco "with".

O metodo json.load() da biblioteca json do Python é utilizado para ler dados JSON a partir de um arquivo ou de uma string e convertê-los em objetos Python correspondentes, como dicionários ou listas. Isso permite que você trabalhe com dados JSON de forma mais fácil e eficiente dentro do seu código Python."""
```

```python
# Transformando dados em uma DataFrame
dados_normalizados = pd.json_normalize(json_bruto)
dados_normalizados.head()

"""O metodo json_normalize do pandas é utilizado para "achatar" estruturas de dados JSON aninhadas em um DataFrame plano. Isso é especialmente útil quando você está lidando com dados JSON que contêm listas ou dicionários dentro de outros dicionários, e você deseja transformar esses dados em um formato tabular para facilitar a análise."""
```

Transformação inicial dos dados

```python
# viIsualizando as informações gerais do DataFrame
dados_normalizados.info()
```

```python
# Erro ao tentar converter a coluna 'conta.cobranca.Total' para float64 devido a presença de valores não numéricos
# Convertendo a coluna 'conta.cobranca.Total' para float64, tratando erros
dados_normalizados['conta.cobranca.Total'] = dados_normalizados['conta.cobranca.Total'].astype('float64')
```

```python
# mostrando os índices dos registros que não foram convertidos na coluna 'conta.cobranca.Total'
idx = dados_normalizados[dados_normalizados['conta.cobranca.Total'] == ' '].index
idx
```

```python
# corrigindo os valores da coluna 'conta.cobranca.Total' atribuindo o valor 24 diretamente  
dados_normalizados.loc[idx, "cliente.tempo_servico"] = 24
```

```python
# verificando se os valores foram corrigidos
dados_normalizados.loc[idx][['cliente.tempo_servico', 'conta.contrato', 'conta.cobranca.mensal', 'conta.cobranca.Total']]
```

```python
# Convertendo novamente a coluna 'conta.cobranca.Total' para float64, agora sem erros
dados_normalizados['conta.cobranca.Total'] = dados_normalizados['conta.cobranca.Total'].astype('float64')
dados_normalizados.info()
```

```python
# visualizando os valores únicos de cada coluna do DataFrame
for col in dados_normalizados.columns:
    print(f"Coluna: {col}")
    print(dados_normalizados[col].unique())
    print("-" * 30)
    
    """O metodo unique() do pandas é utilizado para obter os valores únicos presentes em uma série ou coluna de um DataFrame. Ele retorna um array numpy contendo todos os valores distintos encontrados na série, eliminando quaisquer duplicatas. Isso é útil para entender a diversidade dos dados em uma coluna específica e para identificar categorias ou valores distintos presentes nos dados."""
```

```python
# visualizando os registros com valores vazios na coluna 'Churn'
dados_normalizados.query('Churn == ""')


""""O metodo query() do pandas é utilizado para filtrar linhas de um DataFrame com base em uma expressão booleana. Ele permite que você escreva consultas de forma mais legível e concisa, utilizando uma sintaxe semelhante à SQL. A expressão dentro do query() deve ser uma string que representa a condição que as linhas devem satisfazer para serem incluídas no resultado."""
```

```python
# Criando um novo DataFrame sem os registros com valores vazios na coluna 'Churn' e copiando os dados
dados_sem_vazios = dados_normalizados[dados_normalizados['Churn'] != ''].copy()
dados_sem_vazios
```

```python
# Verificando que há indexes faltando no novo DataFrame
dados_sem_vazios.info()
```

```python
# resetando os índices do DataFrame sem valores vazios
dados_sem_vazios.reset_index(drop=True, inplace=True)
dados_sem_vazios
```

```python
dados_sem_vazios.duplicated()

"""O metodo duplicated() do pandas é utilizado para identificar linhas duplicadas em um DataFrame. Ele retorna uma série booleana indicando se cada linha é uma duplicata de uma linha anterior. Isso é útil para detectar e remover registros duplicados em conjuntos de dados, garantindo a integridade e a qualidade dos dados para análise."""
```

```python
# Removendo os registros duplicados do DataFrame sem valores vazios
dados_sem_vazios.drop_duplicates(inplace=True)


"""O metodo drop_duplicates() do pandas é utilizado para remover linhas duplicadas de um DataFrame. Ele identifica e elimina registros que são idênticos em todas as colunas ou em colunas específicas, mantendo apenas a primeira ocorrência de cada conjunto de duplicatas. Isso é útil para garantir a integridade dos dados e evitar redundâncias em análises e operações subsequentes.

implace=True significa que a operação será realizada no próprio DataFrame, modificando-o diretamente, em vez de retornar uma nova cópia do DataFrame com as duplicatas removidas."""
```

```python
# Verificando o número de registros duplicados após a remoção
dados_sem_vazios.duplicated().sum()
```

Identificando e Substituindo dados nulos

```python
# Identificando e Substituindo dados nulos
dados_sem_vazios.isna()

"""O metodo isna() do pandas é utilizado para identificar valores nulos ou ausentes em um DataFrame ou Série. Ele retorna uma estrutura booleana do mesmo formato, onde cada elemento é True se o valor correspondente for nulo (NaN) e False caso contrário. Isso é útil para detectar e lidar com dados faltantes em conjuntos de dados, permitindo que os usuários tomem decisões informadas sobre como tratar esses valores em análises subsequentes."""
```

```python
# Contando o número de valores nulos em cada coluna do DataFrame sem valores vazios
dados_sem_vazios.isna().sum()
```

```python
# Contando o número total de valores nulos no DataFrame sem valores vazios  
dados_sem_vazios.isna().sum().sum()
```

```python
# Visualizando os registros com valores nulos
dados_sem_vazios[dados_sem_vazios.isna().any(axis=1)]

"""O metodo any(axis=1) é utilizado para verificar se há pelo menos um valor True em cada linha (quando axis=1) ou em cada coluna (quando axis=0) de um DataFrame ou Série. No contexto de dados nulos, ele pode ser usado para identificar linhas ou colunas que contêm valores nulos, retornando True para aquelas que possuem pelo menos um valor nulo e False para aquelas que não possuem nenhum valor nulo."""
```

```python
# Criando um filtro para identificar os registros com valores nulos na coluna 'cliente.tempo_servico'
filtro = dados_sem_vazios['cliente.tempo_servico'].isna()
```

```python
# Visualizando os registros com valores nulos na coluna 'cliente.tempo_servico' 
dados_sem_vazios[filtro][['cliente.tempo_servico', 'conta.cobranca.mensal', 'conta.cobranca.Total']]
```

```python
# Substituindo os valores nulos na coluna 'cliente.tempo_servico' pela divisão arredondada para cima da coluna 'conta.cobranca.Total' pela coluna 'conta.cobranca.mensal'
dados_sem_vazios['cliente.tempo_servico'].fillna(
    np.ceil(dados_sem_vazios['conta.cobranca.Total'] / dados_sem_vazios['conta.cobranca.mensal']), inplace=True
)

"""O metodo fillna() do pandas é utilizado para preencher valores nulos (NaN) em um DataFrame ou Série com um valor específico ou utilizando uma estratégia de preenchimento, como o valor anterior ou posterior. Isso é útil para lidar com dados faltantes, permitindo que os usuários substituam esses valores nulos por informações relevantes, facilitando análises e operações subsequentes. O parâmetro inplace=True indica que a operação será realizada no próprio DataFrame, modificando-o diretamente, em vez de retornar uma nova cópia do DataFrame com os valores nulos preenchidos."""
```

```python
# Visualizando os registros com valores nulos na coluna 'cliente.tempo_servico' após a substituição
dados_sem_vazios[filtro][['cliente.tempo_servico', 'conta.cobranca.mensal', 'conta.cobranca.Total']]
```

```python
# Visualizando o número de valores nulos em cada coluna do DataFrame sem valores vazios após a substituição
dados_sem_vazios.isna().sum()


"""Existem alguns métodos da biblioteca Pandas que nos auxiliam a identificar quais são dados nulos em nosso DataFrame, como:

isnull(): retorna um DataFrame de valores booleanos, no qual True indica o valor nulo e False indica o valor válido.
notnull(): retorna um DataFrame de valores booleanos, no qual True indica o valor válido e False indica o valor nulo.
isna(): retorna um DataFrame de valores booleanos, com True indicando o valor nulo e False o valor válido (a mesma funcionalidade do método isnull()).
notna(): retorna um DataFrame de valores booleanos, em que True indica o valor válido e False indica o valor nulo (a mesma funcionalidade do método notnull()).

Além disso, podemos utilizar o método fillna() para substituir os valores nulos por um valor específico ou por uma estratégia de preenchimento, como o valor anterior ou posterior. Isso é útil para lidar com dados faltantes e garantir a integridade dos dados para análises subsequentes.
# Identificando e Substituindo dados nulos"""
```

Retirando o restante dos dados nulos do DataFrame

```python
# Visualizando a distribuição de valores da coluna 'conta.contrato'
dados_normalizados['conta.contrato'].value_counts()
```

```python
# Atribuindo o valor das colunas a serem removidas a variável colunas_dropar
colunas_dropar = ['conta.contrato', 'conta.faturamente_eletronico', 'conta.metodo_pagamento']
```

```python
# Vizualizando os dados nulos nas colunas a serem removidas e somando a quantidade de registros com valores nulos   
dados_sem_vazios[colunas_dropar].isna().any(axis=1).sum()

"""O metodo any(axis=1) é utilizado para verificar se há pelo menos um valor True em cada linha (quando axis=1) ou em cada coluna (quando axis=0) de um DataFrame ou Série. No contexto de dados nulos, ele pode ser usado para identificar linhas ou colunas que contêm valores nulos, retornando True para aquelas que possuem pelo menos um valor nulo e False para aquelas que não possuem nenhum valor nulo.

O metodo isna() é utilizado para verificar se todos os valores em um DataFrame ou Série são iguais. Ele retorna True se todos os valores forem iguais e False se houver pelo menos um valor diferente."""
```

```python
# Retirando o restante dos dados nulos do DataFrame e atribuindo a variável dados_sem_nulos e copiando os dados
df_dados_sem_nulos = dados_normalizados.dropna(subset=colunas_dropar).copy()
df_dados_sem_nulos.head()
```

```python
# Resetando o index do DataFrame sem nulos
df_dados_sem_nulos.reset_index(drop=True, inplace=True)
df_dados_sem_nulos.head()
```

```python
# Visualizando a contagem de valores nulos no DataFrame sem nulos
df_dados_sem_nulos.isna().sum()
```
