```python
# transformando o DataFrame de formato largo (wide) para formato longo (long) usando o método melt().
emissoes_por_ano = emissoes_gases.melt(id_vars = colunas_info, value_vars = colunas_emissao, var_name = 'Ano', value_name = 'Emissão')
```

```python
"""Os termos wide e long são usados para descrever a organização dos dados em um banco de dados relacional.
No formato wide, também conhecido como tabelas de pivô, as tabelas têm uma única linha de cabeçalho contendo diversas colunas, geralmente mais colunas do que linhas, e são úteis para armazenar informações que possuem muitos atributos, como questionários e pesquisas. As colunas podem conter valores ausentes para as células em que os dados não se aplicam.
No formato long, também conhecido como formato normalizado ou estreito, as tabelas armazenam cada observação em uma linha separada. Ao invés de ter uma única linha de cabeçalho contendo todas as colunas, as tabelas no formato long geralmente têm duas colunas principais de informação, uma para os nomes das variáveis e outra para os valores das variáveis. São úteis para armazenar informações com muitas observações, como séries temporais ou dados de experimentos científicos.

Na biblioteca pandas, é possível fazer uma mudança do formato wide para o formato long com o uso do método melt(). A sintaxe para utilizar o método melt() é a seguinte: 
dataframe.melt(id_vars = [...], value_vars = [...], var_name = '..', value_name = '..')

O parâmetro id_vars é utilizado para identificar os nomes das colunas que serão mantidas no momento da transformação. Já o parâmetro value_vars é utilizado para identificar as colunas que serão transformadas nas colunas principais do formato long, uma contendo as categorias com os nomes das colunas e outra contendo os valores. O parâmetro var_name é usado para nomear a coluna contendo as categorias e o parâmetro value_name para nomear a coluna com os valores numéricos.
"""
```

EMISSÃO DE GASES POR SETOR

```python
# Agrupando as emissões por gás e setor
gas_por_setor = emissoes_por_ano.groupby(['Gás', 'Nível 1 - Setor'])['Emissão'].sum()
gas_por_setor
```

```python
# exibindo as emissões do gás CO2 por setor
gas_por_setor.xs('CO2 (t)',level=0)

"""O metodo xs() do pandas é utilizado para selecionar dados em um DataFrame ou Series com MultiIndex (índice hierárquico). Ele permite acessar um nível específico do índice, retornando uma fatia dos dados correspondentes a esse nível. No exemplo acima, gas_por_setor é uma Series com MultiIndex onde o primeiro nível é 'Gás' e o segundo nível é 'Nível 1 - Setor'. Ao chamar gas_por_setor.xs('CO2 (t)', level=0), estamos selecionando todas as linhas onde o valor do primeiro nível do índice ('Gás') é igual a 'CO2 (t)'. O parâmetro level=0 indica que estamos nos referindo ao primeiro nível do índice. O resultado é uma Series contendo as emissões do gás CO2 por setor."""
```

```python
# exibindo as emissões do gás CO2 no setor de Mudança de Uso da Terra e Floresta
gas_por_setor.xs(('CO2 (t)', 'Mudança de Uso da Terra e Floresta'), level=[0,1])
```

```python
# exibindo o setor com maior emissão do gás CO2
gas_por_setor.xs('CO2 (t)',  level=0).idxmax()

"""O metodo idxmax() do pandas é utilizado para encontrar o índice (rótulo) do valor máximo em uma Series ou DataFrame. No exemplo acima, gas_por_setor.xs('CO2 (t)', level=0) retorna uma Series contendo as emissões do gás CO2 por setor. Ao chamar o método idxmax() nessa Series, estamos obtendo o índice (rótulo) do setor que possui a maior emissão de CO2. O resultado é o nome do setor com a maior emissão de CO2."""
```

```python
"""Um DataFrame multi Index é uma estrutura de dados em pandas que permite a criação de um índice hierárquico, isto é, com um índice que possui múltiplos níveis. Isso permite a organização dos dados de maneira mais detalhada do que um índice simples"""
```

```python
# exibindo o setor com maior emissão de cada gás
gas_por_setor.groupby(level=0).idxmax()
```

```python
# armazenando os valores máximos de emissão de cada gás em uma variável
valores_max = gas_por_setor.groupby(level=0).max().values
```

```python
# criando uma tabela sumarizada com os setores de maior emissão por gás e seus respectivos valores
tabela_sumarizada = gas_por_setor.groupby(level=0).idxmax().to_frame()
tabela_sumarizada.insert(1, 'Quantidade de emissão', valores_max)
tabela_sumarizada

"""A tabela sumarizada criada acima apresenta os setores com maior emissão para cada gás, juntamente com a quantidade de emissão correspondente. A primeira coluna da tabela contém os índices dos setores com maior emissão para cada gás, enquanto a segunda coluna exibe os valores máximos de emissão associados a esses setores. Essa tabela fornece uma visão clara e concisa dos principais contribuintes para as emissões de gases estufa em diferentes setores.

A função insert() do pandas é utilizada para adicionar uma nova coluna em um DataFrame em uma posição específica. No exemplo acima, tabela_sumarizada é um DataFrame que contém os setores com maior emissão para cada gás. Ao chamar tabela_sumarizada.insert(1, 'Quantidade de emissão', valores_max), estamos inserindo uma nova coluna chamada 'Quantidade de emissão' na posição 1 (segunda coluna) do DataFrame, e os valores dessa coluna são fornecidos pela variável valores_max, que contém os valores máximos de emissão para cada gás. O resultado é um DataFrame atualizado com a nova coluna adicionada.

O metodo .to_frame() do pandas é utilizado para converter uma Series em um DataFrame. No exemplo acima, gas_por_setor.groupby(level=0).idxmax() retorna uma Series contendo os índices dos setores com maior emissão para cada gás. Ao chamar o método to_frame() nessa Series, estamos convertendo-a em um DataFrame, onde a Series original se torna uma coluna do DataFrame. Isso permite que possamos manipular e analisar os dados de forma mais estruturada, utilizando as funcionalidades disponíveis para DataFrames no pandas."""
```

```python
# exibindo o setor com maior emissão de cada gás, após trocar os níveis do índice
gas_por_setor.swaplevel(0,1)

"""O metodo swaplevel() do pandas é utilizado para trocar os níveis de um MultiIndex (índice hierárquico) em um DataFrame ou Series. No exemplo acima, gas_por_setor é uma Series com MultiIndex onde o primeiro nível é 'Gás' e o segundo nível é 'Nível 1 - Setor'. Ao chamar gas_por_setor.swaplevel(0,1), estamos trocando a posição dos níveis do índice, fazendo com que o segundo nível ('Nível 1 - Setor') se torne o primeiro nível e o primeiro nível ('Gás') se torne o segundo nível. O resultado é uma Series com os níveis do índice invertidos."""
```

```python
# exibindo o setor com maior emissão de cada gás, após trocar os níveis do índice
gas_por_setor.swaplevel(0,1).groupby(level=0).idxmax()
```
