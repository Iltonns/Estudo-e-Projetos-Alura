```python
import pandas as pd
dados = pd.read_csv(r"C:\Users\eleil\OneDrive - Uniasselvi\Alura\Python para Data Sciense\Pandas\aluguel.csv", sep = ";")
dados.head()
```

```python
# Calcular a média de quartos dos imóveis   
media_quartos = dados['Quartos'].mean()
print(f"A média de quartos dos imóveis é: {media_quartos:.2f}")
```

```python
# Calcular a quantidade de bairros únicos na base de dados
Bairros_unicos = dados['Bairro'].unique()
quantidade_bairros = len(Bairros_unicos)
print(f"A quantidade de bairros únicos é: {quantidade_bairros}")
```

```python
dados.groupby('Tipo')[['Valor']].mean().sort_values("Valor")

```

```python
aluguel_elevado = dados.groupby('Tipo')[['Valor']].mean().sort_values("Valor", ascending = False).head()
aluguel_elevado.plot(kind='barh', figsize=(14,10), color='blue');


```
